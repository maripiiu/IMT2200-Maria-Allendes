Existe una gran diversidad de categorías de juegos de mesa. Ahora nos concentraremos en un grupo específico de ellas, con el fin de analizar cómo a cambiado la cantidad de juegos de estas clases desde 1980 hasta hoy.

**a) (0.7 pts)** ¿Cuáles son las 5 categorías más comunes en los juegos del dataset? Muestre la cantidad de juegos que hay de cada una. Puede graficar estos valores, o bien, entregar un DataFrame con sus valores.

Las 5 categorías mas comunes en los juegos del data set son: 1002 - 1041 - 1019 - 1030 - 1017, cabe destacar que esto siendo en categorías individuales puesto que habían juegos que contenían mas de una categoría como por ejemplo: CATAN  1021,1026, por lo que para este ejercicio separé las categorias para tener una mejor vista de los datos :)

```python
#Se asume que una categoría = xxxx entiendase como un n° de 4 dígitos por lo que en un juego puede haber mas de 1 categoría
top_5_clean = fd.dropna(subset = ["category"]).copy()
top_5_clean["category"] = top_5_clean["category"].astype(str)
top_split = top_5_clean["category"].str.split(",", expand=True)
#print(top_split)
```

```python
todas_categorias = []
for columna in top_split.columns:
    categ_columna = top_split[columna].dropna()
    todas_categorias.extend(categ_columna.tolist())
categorias_series = pd.Series(todas_categorias).value_counts()
top_5_categ = categorias_series.sort_values(ascending = False).head(5)
#df
top_5_df = categorias_series.head(5).reset_index()
top_5_df.columns = ["Categoría", "Cantidad"]
print("TOP 5 Juegos Categorías Individuales:")
print()
print(top_5_df)
```

```python
plt.figure(figsize=(12,6))
categorias_str = top_5_df["Categoría"].astype(str)
bars = plt.bar(categorias_str, top_5_df["Cantidad"], color =['#FF6B6B', '#4ECDC4', '#7ddeff', '#9b84ff', '#ffe37e'])
plt.xlabel("Codigos de Categoría de Juegos", fontsize=11)
plt.ylabel("Cantidad de Juegos", fontsize=11)
plt.title("TOP 5: Categorias mas Comunes en Juegos de Mesa ", fontsize = 12);
```
